```python
from pathlib import Path

project_root = Path.cwd()

print("Project folder:")
print(project_root)

print("\nRaw data folders:")
for folder in [
    "data/raw/psa",
    "data/raw/pagasa",
    "data/raw/doe",
    "data/raw/cpi"
]:
    path = project_root / folder
    print(f"\n{folder}:")
    
    files = list(path.iterdir())
    
    if files:
        for file in files:
            print("  -", file.name)
    else:
        print("  (empty)")
```
